# 1) Recap of lecture
**Copy and paste** your code from lecture for the neural network with the **default** settings (i.e. before we improved the model). Make sure to also import the `data.py` file. 

```python
# Your code here
%pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cpu
import data

import matplotlib.pyplot as plt

plt.figure(figsize=(3,3))
plt.scatter(data.x_values, data.y_values, c=data.color_values)
plt.show()

import torch

loss_fn = torch.nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

model = torch.nn.Sequential(
    torch.nn.Linear(2, 50),
    torch.nn.Sigmoid(), 
    torch.nn.Linear(50, 1)
)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

for t in range(1000):
    y_pred = model(data.x)           
    loss = loss_fn(y_pred, data.y)   
    optimizer.zero_grad()             
    loss.backward()                 
    optimizer.step()              

    if t % 100 == 0 or t%999 ==0:
        print(f"Iteration {t}, Loss: {loss.item()}")

new_x = torch.linspace(-10, 10, 100) # 1D array of values
grid = torch.stack(torch.meshgrid((new_x, new_x)), dim=2) # Making our y, meshing them together

output = model(grid).detach() # just evalutating, taking the x/y pairs and predicting a "z"

output = output.squeeze(2).t() # Making it the right dimension
plt.imshow(output, extent=(-10,10,-10,10))
plt.show()   
plt.imshow(data.pattern)
plt.show()
```

What is the difference between artificial intelligence, machine learning and deep learning?

```python
# Your comment here 
'''
deep learning is a subset of machine learning which is a subset of AI. 
AI is the idea that we can teach machines to do a bunch of tasks that normally require human intelligence
ML is putting a bunch of data into a model and then letting the machine complete tasks using that data after it has "learned"
Deep Learning is an advanced version of ml where lots of data is required and requires some networks to analyze the data to find patterns
'''
```

What is a neural network?

```python
# Your comment here
'''
A neural network is basically an algorithm where computers can process data and extrapolate data using "neurons" to reveal a pattern 
'''
```

In lecture, we used `torch.nn.Linear(2, 50)`. What do the numbers `2` and `50` represent? 

```python
# Your comment here
'''
2 is the number of inputs (ex. x,y)
50 is the number of output features (kinda like the number of neurons)
'''
```

What does the activation function do?

```python
# Your comment here
'''
We used torch.nn.Sigmoid() as the activation function which causes the model to not necessarily be linear since linear often limits
what pytorch can do. This allows the neurons to be more continuous and smooth.
'''
```

What is a loss function? Why do we need one?

```python
# Your comment here
'''
A loss functions tell us how wrong/inaccurate the model is. This is so we know what can be fixed and find better values so that
there is less loss which brings us closer to the real pattern.
'''
```

Why do we call `optimizer.zero_grad()` before calling `loss.backward()`?

```python
# Your comment here
'''
We call optimizer.zero_grad() so the gradients are reset so then the loss.backward() can recompute the gradients for the updated model
'''
```

With your default code above, run your model with **30 neurons**. Compare the loss functions from lecture to your modification on the **1000th** iteration. 

```python
# Your comment here
plt.figure(figsize=(3,3))
plt.scatter(data.x_values, data.y_values, c=data.color_values)
plt.show()

import torch

loss_fn = torch.nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

model = torch.nn.Sequential(
    torch.nn.Linear(2, 30),
    torch.nn.Sigmoid(), 
    torch.nn.Linear(30, 1)
)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

for t in range(1000):
    y_pred = model(data.x)           
    loss = loss_fn(y_pred, data.y)   
    optimizer.zero_grad()             
    loss.backward()                 
    optimizer.step()              

    if t % 100 == 0 or t%999==0:
        print(f"Iteration {t}, Loss: {loss.item()}")

new_x = torch.linspace(-10, 10, 100) # 1D array of values
grid = torch.stack(torch.meshgrid((new_x, new_x)), dim=2) # Making our y, meshing them together

output = model(grid).detach() # just evalutating, taking the x/y pairs and predicting a "z"

output = output.squeeze(2).t() # Making it the right dimension
plt.imshow(output, extent=(-10,10,-10,10))
plt.show()   
plt.imshow(data.pattern)
plt.show()
'''
Loss = 0.40 for 30 neurons
Loss = 0.347 for 50 neurons
this makes sense cuz more neurons = better models
'''
```

Return to **50 neurons**. Change your activation function from `torch.nn.Sigmoid()` to `torch.nn.ReLU`. Compare the loss functions from lecture to your modification on the **1000th** iteration. 

```python
# Your comment here
plt.figure(figsize=(3,3))
plt.scatter(data.x_values, data.y_values, c=data.color_values)
plt.show()

loss_fn = torch.nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

model = torch.nn.Sequential(
    torch.nn.Linear(2, 50),
    torch.nn.ReLU(), 
    torch.nn.Linear(50, 1)
)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

for t in range(1000):
    y_pred = model(data.x)           
    loss = loss_fn(y_pred, data.y)   
    optimizer.zero_grad()             
    loss.backward()                 
    optimizer.step()              

    if t % 100 == 0 or t%999==0:
        print(f"Iteration {t}, Loss: {loss.item()}")

new_x = torch.linspace(-10, 10, 100) # 1D array of values
grid = torch.stack(torch.meshgrid((new_x, new_x)), dim=2) # Making our y, meshing them together

output = model(grid).detach() # just evalutating, taking the x/y pairs and predicting a "z"

output = output.squeeze(2).t() # Making it the right dimension
plt.imshow(output, extent=(-10,10,-10,10))
plt.show()   
plt.imshow(data.pattern)
plt.show()
'''
Loss function is significantly less. 0.30 for this vs Sigmoid which was about 0.347
'''
```

What is the difference between `torch.nn.Sigmoid()`, `torch.nn.ReLU` and `torch.nn.Tanh`? 

```python
# Your comment here 
'''
These are all different activation functions but have different models. Sigmoid maps inputs from 0 to 1, Tanh maps inputs from -1 to 1,
while ReLU maps inputs from to to input value. 
'''
```
