## 2 Plotting
Make a scatter plot of **exoplanet Earth radius vs. equilibrium temperatures**. Include units, x/y axis labels, and a legend.

```python
# Your code here
import matplotlib.pyplot as plt
df.plot.scatter(x='pl_rade', y='pl_eqt')
plt.xlabel('Exoplanet Radius (R$_E$)')
plt.ylabel('Equilibrium Temperatures (K)')
plt.legend(['Planets'], loc='lower right')
plt.show()
```

```python
# Your comment here
# In general, as exoplanet radius increases, equilibrium temperature usually also increases. However, there are a few data points
# which do not necessarily follow this trend. This is a positive correlation.
```

## 3 A "Cool" Planet
What planet has the **lowest** equilibrium temperature? Print the planet name and temperature with your `planet_data` variable.

```python
# Your code here
temperatures = list(df['pl_eqt'])
lowest = min(temperatures)
planet_index = temperatures.index(lowest)
planet_names = df['pl_name']
planet_name = planet_names[planet_index]
print(planet_name)
```

Make a new variable that only contains the information for that planet. \
\
*Hint: Use the `.iloc[[#]]` function. Replace # with the proper index.*

```python
# Your code here
planet_info = df.iloc[[planet_index]]
print(planet_info)
```

Print the number of columns in the row. 

```python
# Your code here
print(len(df.loc[planet_index]))
```

Make a new variable that contains only the columns WITHOUT `Nan` values. *Hint: You can use the function `.dropna(axis=1)`.*

```python
# Your code here
planet_info_noNan = planet_info.dropna(axis=1)
print(planet_info_noNan)
```

Print how many columns you have now.

```python
# Your code here
print(len(list(planet_info_noNan)))
```
