# 2 Plotting a Function
Goal: Write a **function** that will plot $f(x) = x^2$ and $g(x) = \sqrt{x}$ on the same plot and return the $f(x)$ and $g(x)$ values as separate lists. Your plot needs to have the following:
- An appropriate figure size (`figsize=(10,8)` is recommended)
- Labeled x and y axis
- A title
- Different colors and linestyles for each function
- A legend
- Your domain should be $x \in [0, 10]$. *Hint: Use `np.linspace` to set a range of values for your functions.*

Your function should return:
- A list of $f(x)$ values
- A list of $g(x)$ values.

```python
# Your code here
def plot_two_functions():
    """
    Inputs: none
    Outputs: none
    Shows a graph of y vs x^2 and y vs sqrt(x) using matplotlib
    """
    fig = plt.figure(figsize=(10,8))
    x = np.linspace(0,10)
    y = x**2
    y2 = x**0.5 # This is just another way of saying sqrt(x) although you could use math.sqrt()
    plt.title("Y vs X^2 and Y vs sqrt(x)")
    plt.xlabel("X-axis")
    plt.ylabel("Y-axis")
    plt.plot(x,y, label = "f(x) = x^2", color = "blue", linestyle="dotted")
    plt.plot(x,y2, label = "g(x) = sqrt(x)", color = "green", linestyle="dashed")
    plt.legend()
    plt.show()
    return x,y
plot_two_functions()
```

```python
# Import your module and/or your code here
import math # required to have math.pi
def three_by_three():
    """
    Inputs: none
    Outputs: none
    Shows a 3x3 grid of subplots that shows 9 trig functions
    """
    fig = plt.figure(figsize=(18,18))
    fig, axes = plt.subplots(3, 3)
    x = np.linspace(-2*math.pi,2*math.pi)
    cos=np.cos(x)
    sin=np.sin(x)
    tan=np.tan(x)
    arccos=np.arccos(x)
    arcsin=np.arcsin(x)
    arctan=np.arctan(x)
    sec=1/np.cos(x) # These three lines give the 1/basic trig functions since numpy does not have a builtin sec, csc, or cot function.
    csc=1/np.sin(x)
    cot=1/np.tan(x)
    axes[0,0].plot(x,cos, label = "f(x)=cos(x)", color = "blue", linestyle="solid")
    axes[0,1].plot(x,sin, label = "f(x)=sin(x)", color="green", linestyle="solid")
    axes[0,2].plot(x,tan, label = "f(x)=tan(x)", color="red", linestyle="solid")
    axes[1,0].plot(x,arccos, label = "f(x)=arccos(x)", color = "orange", linestyle="solid")
    axes[1,1].plot(x,arcsin, label = "f(x)=arcsin(x)", color="indigo", linestyle="solid")
    axes[1,2].plot(x,arctan, label = "f(x)=arctan(x)", color="maroon", linestyle="solid")
    axes[2,0].plot(x,sec, label = "f(x)=sec(x)", color = "yellow", linestyle="solid")
    axes[2,1].plot(x,csc, label = "f(x)=csc(x)", color="black", linestyle="solid")
    axes[2,2].plot(x,cot, label = "f(x)=cot(x)", color="brown", linestyle="solid")
    axes[0,0].set_title("cos(x) vs x")
    axes[0,1].set_title("sin(x) vs x")
    axes[0,2].set_title("tan(x) vs x")
    axes[1,0].set_title("arccos(x) vs x")
    axes[1,1].set_title("arcsin(x) vs x")
    axes[1,2].set_title("arctan(x) vs x")
    axes[2,0].set_title("sec(x) vs x")
    axes[2,1].set_title("csc(x) vs x")
    axes[2,2].set_title("cot(x) vs x")
    axes[0,0].set_xlabel("X-axis")
    axes[0,0].set_ylabel("Y-axis")
    axes[0,1].set_xlabel("X-axis")
    axes[0,1].set_ylabel("Y-axis")
    axes[0,2].set_xlabel("X-axis")
    axes[0,2].set_ylabel("Y-axis")
    axes[1,0].set_xlabel("X-axis")
    axes[1,0].set_ylabel("Y-axis")
    axes[1,1].set_xlabel("X-axis")
    axes[1,1].set_ylabel("Y-axis")
    axes[1,2].set_xlabel("X-axis")
    axes[1,2].set_ylabel("Y-axis")
    axes[2,0].set_xlabel("X-axis")
    axes[2,0].set_ylabel("Y-axis")
    axes[2,1].set_xlabel("X-axis")
    axes[2,1].set_ylabel("Y-axis")
    axes[2,2].set_xlabel("X-axis")
    axes[2,2].set_ylabel("Y-axis")
    axes[0,0].legend()
    axes[0,1].legend()
    axes[0,2].legend()
    axes[1,0].legend()
    axes[1,1].legend()
    axes[1,2].legend()
    axes[2,0].legend()
    axes[2,1].legend()
    axes[2,2].legend()
    plt.tight_layout()
    plt.show()
three_by_three()
```
